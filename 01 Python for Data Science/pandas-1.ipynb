## Introduction

### What is Pandas?

Pandas is a library for data manipulation and analysis. It provides:

* **DataFrame**: 2D labeled data structure (like Excel spreadsheet)
* **Series**: 1D labeled array
* **Powerful tools**: For cleaning, transforming, and analyzing data

### Why Pandas?

* **Easy data manipulation**: Load, clean, transform data easily
* **Handles missing data**: Built-in functions for dealing with NaN
* **Time series**: Excellent support for time-based data
* **Integration**: Works seamlessly with NumPy, Matplotlib, Scikit-learn

## Series and DataFrames

### Series

A Series is a one-dimensional labeled array.

````python
# Creating Series
s = pd.Series([1,2,3,4,5])
print(s)
````

````python
# With custom index
s = pd.Series([10, 20, 30], index=['a','b','c'])
print(s)
````

````python
# From dictionary
s = pd.Series({
    'Alice':25,
    'Bob':30,
    'Charlie':35
})
print(s)
````

DataFrame

A DataFrame is a 2D labeled data structure with columns of potentially different types.

````python
# Creating DataFrame from dictionary
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'Diana'],
    'Age': [25, 30, 35, 28],
    'City': ['New York', 'London', 'Tokyo', 'Paris'],
    'Salary': [50000, 60000, 70000, 55000]
}

df = pd.DataFrame(data)
print(df)
````

````python
# From list of lists
data = [['Alice', 25], ['Bob', 30], ['Charlie', 35]]
df = pd.DataFrame(data, columns=['Name','Age'])
print(df)
````

````python
# From NumPy array
arr = np.array([[1,2,3], [4,5,6]])
df = pd.DataFrame(arr, columns=['A','B','C'])
print(df)
````

## Grouping and Aggregation

### GroupBy

````python
# Group by column
import pandas as pd
import numpy as np
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'Diana', 'Nafis', 'Mehedi', 'Evan'],
    'Age': [25, 30, 35, 28, 32, 29, 31],
    'City': ['New York', 'London', 'Tokyo', 'Paris', 'New York', 'London', 'New York'],
    'Salary': [50000, 60000, 70000, 55000, 58000, 62000, 59000]
}
df = pd.DataFrame(data)
print(df)
````

````python
# Group by column
grouped = df.groupby('City')

# Apply aggregation functions
city_states = grouped.agg({
    'Age': 'mean',  # Average age per city
    'Salary': ['mean', 'sum', 'count']
})
print(city_states)
````

```python
# Common aggregations
grouped.mean()  # Average of numeric columns per group
grouped.sum()  # Sum of numeric columns per group
grouped.count() # Count of non-null values per group
grouped.size()  # Size of each group (including NaN)
grouped.min()   # Minimum value per group
grouped.max()   # Maximum value per group
grouped.std()   # Standard deviation per group
grouped.describe()  # Statistical summary per group

### Custom Aggregations

````python
# Multiple aggregations
agg_dict = {
    'Age': ['mean', 'min', 'max'],  # Multiple aggregations for Age
    'Salary': ['mean', 'sum', 'count']  # Multiple aggregations for Salary
}
result = df.groupby('City').agg(agg_dict) 
print(result)
````

````python
# Custom function
def range_func(x):
    return x.max() - x.min()
result = df.groupby('City')['Age'].agg(range_func)
print(result)
````

## Merging and Joining
### Merge (SQL-like joins)

````python
# Sample DataFrames
df1 = pd.DataFrame({
    'ID': [1, 2, 3],
    'Name': ['Alice', 'Bob', 'Charlie']
})
print(df1)
print('--'*30)

df2 = pd.DataFrame({
    'ID': [2, 3, 4],
    'Salary': [60000, 70000, 80000]
})
print(df2)
print('--'*30)

# Inner join (default)
merged = pd.merge(df1, df2, on='ID', how='inner')
print(merged)
print('--'*30)

# left Join
merged_left = pd.merge(df1, df2, on='ID', how='left')
print(merged_left)
print('--'*30)

# Right join
merged_right = pd.merge(df1, df2, on='ID', how='right')
print(merged_right)
print('--'*30)

# Outer join
merged_outer = pd.merge(df1, df2, on='ID', how='outer')
print(merged_outer)
````

### Concatenation

````python
# Concatenate DataFrames
df1 = pd.DataFrame({'A':[1,2], 'B':[3,4]})
df2 = pd.DataFrame({'A':[5,6], 'B':[7,8]})
print(df1)
print('--'*30)
print(df2)
print('--'*30)


# Vertical concatenation
result = pd.concat([df1, df2], axis=0)  # axis=0 for vertical
print('Vertical concat:\n', result)
print('--'*30)

# Horizontal concatenation
result = pd.concat([df1, df2], axis=1)  # axis=1 for horizontal
print('Horizontal concat:\n', result)
print('--'*30)

# With keys
result = pd.concat([df1, df2], keys=['first', 'second'])
print('Concatenation with keys:\n', result)
````

### Rolling & Expanding Windows
Calculate moving averages, cumulative sums, or trends over time:
# 🔥 Time Series Cheat Sheet

## 📊 Core Concepts

| Concept        | Meaning (Bangla)                      | Simple Idea |
|----------------|--------------------------------------|-------------|
| Rolling        | fixed window sliding average         | window slide করে calculation |
| Expanding      | শুরু থেকে এখন পর্যন্ত সব ডেটা        | cumulative calculation |
| Shift          | আগের/পরের value দেখানো              | lag / lead data |
| pct_change     | percentage difference                | growth rate (%) |

---

## 🎯 One Line Summary

- 👉 **Rolling** = sliding window (moving average)
- 👉 **Expanding** = growing window (start → now)
- 👉 **Shift** = time shift (past/future value)
- 👉 **pct_change** = percentage growth between periods

---

## 🧠 Quick Memory Trick

- 📉 Rolling → "last N days focus"
- 📈 Expanding → "all history till now"
- ⏪ Shift → "move data forward/backward"
- 📊 pct_change → "how much changed in %"

```python

# Rolling window (moving average)
df['Rolling_mean_7'] = df['Value'].rolling(window=7).mean()  # 7-day rolling mean, প্রতিদিন শেষ ৭ দিনের গড় বের করবে
df['Rolling_std_15'] = df['Value'].rolling(window=15).std()   # 15 দিনের volatility (standard deviation)
df['Rolling_sum_30'] = df['Value'].rolling(window=30).sum()   # 30 দিনের total sales
df['Rolling_max_90'] = df['Value'].rolling(window=90).max()   # 90 দিনের highest value

# Rolling with minimum periods
df['Rolling_mean_3_min_periods'] = df['Value'].rolling(window=3, min_periods=3).mean()  # minimum 3 data থাকলেই calculation হবে

# Expanding window (cumulative) - “শুরু থেকে এখন পর্যন্ত সব ডেটা নিয়ে হিসাব করা”
df['Expanding_mean'] = df['Value'].expanding().mean()  # Cumulative mean from start to current point
df['Expanding_sum'] = df['Value'].expanding().sum()    # Cumulative sum from start to current point
df['Expanding_max'] = df['Value'].expanding().max()    # এখন পর্যন্ত সর্বোচ্চ মান, Cumulative max from start to current point
df['Expanding_min'] = df['Value'].expanding().min()    # এখন পর্যন্ত সর্বনিম্ন মান, Cumulative min from start to current point

# Shift values (lag/lead) - “ডেটাকে উপরে বা নিচে সরানো”
df['Previous'] = df['Value'].shift(1)  # Previous day’s value (lag of 1)
df['Next'] =df['Value'].shift(-1)     # Next day’s value (lead of 1)
df['Lag_7'] = df['Value'].shift(7)    # 7-day lagged value- ৭ দিন আগের value

# Percentage change - “এক দিন থেকে পরের দিনে কত % change হয়েছে”
df['pct_change_1'] = df['Value'].pct_change() # “এক দিন থেকে পরের দিনে কত % change হয়েছে”
df['pct_change_7'] = df['Value'].pct_change(period=7)  # 7-day percentage change - ৭ দিন আগের থেকে এখন পর্যন্ত কত % change হয়েছে
