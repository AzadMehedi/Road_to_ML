## Introduction

### What is Pandas?

Pandas is a library for data manipulation and analysis. It provides:

* **DataFrame**: 2D labeled data structure (like Excel spreadsheet)
* **Series**: 1D labeled array
* **Powerful tools**: For cleaning, transforming, and analyzing data

### Why Pandas?

* **Easy data manipulation**: Load, clean, transform data easily
* **Handles missing data**: Built-in functions for dealing with NaN
* **Time series**: Excellent support for time-based data
* **Integration**: Works seamlessly with NumPy, Matplotlib, Scikit-learn

## Series and DataFrames

### Series

A Series is a one-dimensional labeled array.

````python
# Creating Series
s = pd.Series([1,2,3,4,5])
print(s)
````

````python
# With custom index
s = pd.Series([10, 20, 30], index=['a','b','c'])
print(s)
````

````python
# From dictionary
s = pd.Series({
    'Alice':25,
    'Bob':30,
    'Charlie':35
})
print(s)
````

DataFrame

A DataFrame is a 2D labeled data structure with columns of potentially different types.

````python
# Creating DataFrame from dictionary
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'Diana'],
    'Age': [25, 30, 35, 28],
    'City': ['New York', 'London', 'Tokyo', 'Paris'],
    'Salary': [50000, 60000, 70000, 55000]
}

df = pd.DataFrame(data)
print(df)
````

````python
# From list of lists
data = [['Alice', 25], ['Bob', 30], ['Charlie', 35]]
df = pd.DataFrame(data, columns=['Name','Age'])
print(df)
````

````python
# From NumPy array
arr = np.array([[1,2,3], [4,5,6]])
df = pd.DataFrame(arr, columns=['A','B','C'])
print(df)
````

````python
# Boolean indexing
young = df[df['Age']<30]
print(young)
````

````python
# Multiple conditions
filtered = df[(df['Age']>25) & (df['Salary']>55000)]
print(filtered)
````

### iloc vs loc


````python
# iloc: Integer position-based indexing
print(df.iloc[0, 1])  # Row 0, Column 1
print('--'*20)
print(df.iloc[:3, 1:3])   # Rows 0-2, Columns 1-2
print('--'*20)

# loc: Label-based indexing
print(df.loc[0, 'Age'])   # Row with index 0, column 'Age'
print('--'*20)
print(df.loc[:2, 'Name':'City'])   # Rows 0-2, columns Name to City
````
